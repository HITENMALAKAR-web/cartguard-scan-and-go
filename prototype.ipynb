```python
cap = cv2.VideoCapture(0)
prices = {
    '100_milk': 35,
    '101_milk': 35,
    '102_milk': 35,
    '103_milk': 35,
    '104_milk': 35,
    '105_milk': 35,
    '106_milk': 35,
    '107_milk': 35,
    '108_canned_food': 70,
    '109_canned_food': 70,
    '10_puffed_food': 30,
    '110_canned_food': 70,
    '111_canned_food': 70,
    '112_canned_food': 70,
    '113_canned_food': 70,
    '114_canned_food': 70,
    '115_canned_food': 70,
    '116_canned_food': 70,
    '117_canned_food': 70,
    '118_canned_food': 70,
    '119_canned_food': 70,
    '11_puffed_food': 30,
    '120_canned_food': 70,
    '121_canned_food': 70,
    '122_chocolate': 40,
    '123_chocolate': 40,
    '124_chocolate': 40,
    '125_chocolate': 40,
    '126_chocolate': 40,
    '127_chocolate': 40,
    '128_chocolate': 40,
    '129_chocolate': 40,
    '12_puffed_food': 30,
    '130_chocolate': 40,
    '131_chocolate': 40,
    '132_chocolate': 40,
    '133_chocolate': 40,
    '134_gum': 10,
    '135_gum': 10,
    '136_gum': 10,
    '137_gum': 10,
    '138_gum': 10,
    '139_gum': 10,
    '13_dried_fruit': 80,
    '140_gum': 10,
    '141_gum': 10,
    '142_candy': 15,
    '143_candy': 15,
    '144_candy': 15,
    '145_candy': 15,
    '146_candy': 15,
    '147_candy': 15,
    '148_candy': 15,
    '149_candy': 15,
    '14_dried_fruit': 80,
    '150_candy': 15,
    '151_candy': 15,
    '152_seasoner': 50,
    '153_seasoner': 50,
    '154_seasoner': 50,
    '155_seasoner': 50,
    '156_seasoner': 50,
    '157_seasoner': 50,
    '158_seasoner': 50,
    '159_seasoner': 50,
    '15_dried_fruit': 80,
    '160_seasoner': 50,
    '161_seasoner': 50,
    '162_seasoner': 50,
    '163_seasoner': 50,
    '164_personal_hygiene': 100,
    '165_personal_hygiene': 100,
    '166_personal_hygiene': 100,
    '167_personal_hygiene': 100,
    '168_personal_hygiene': 100,
    '169_personal_hygiene': 100,
    '16_dried_fruit': 80,
    '170_personal_hygiene': 100,
    '171_personal_hygiene': 100,
    '172_personal_hygiene': 100,
    '173_personal_hygiene': 100,
    '174_tissue': 60,
    '175_tissue': 60,
    '176_tissue': 60,
    '177_tissue': 60,
    '178_tissue': 60,
    '179_tissue': 60,
    '17_dried_fruit': 80,
    '180_tissue': 60,
    '181_tissue': 60,
    '182_tissue': 60,
    '183_tissue': 60,
    '184_tissue': 60,
    '185_tissue': 60,
    '186_tissue': 60,
    '187_tissue': 60,
    '188_tissue': 60,
    '189_tissue': 60,
    '18_dried_fruit': 80,
    '190_tissue': 60,
    '191_tissue': 60,
    '192_tissue': 60,
    '193_tissue': 60,
    '194_stationery': 30,
    '195_stationery': 30,
    '196_stationery': 30,
    '197_stationery': 30,
    '198_stationery': 30,
    '199_stationery': 30,
    '19_dried_fruit': 80,
    '1_puffed_food': 30,
    '200_stationery': 30,
    '20_dried_fruit': 80,
    '21_dried_fruit': 80,
    '22_dried_food': 60,
    '23_dried_food': 60,
    '24_dried_food': 60,
    '25_dried_food': 60,
    '26_dried_food': 60,
    '27_dried_food': 60,
    '28_dried_food': 60,
    '29_dried_food': 60,
    '2_puffed_food': 30,
    '30_dried_food': 60,
    '31_instant_drink': 100,
    '32_instant_drink': 100,
    '33_instant_drink': 100,
    '34_instant_drink': 100,
    '35_instant_drink': 100,
    '36_instant_drink': 100,
    '37_instant_drink': 100,
    '38_instant_drink': 100,
    '39_instant_drink': 100,
    '3_puffed_food': 30,
    '40_instant_drink': 100,
    '41_instant_drink': 100,
    '42_instant_noodles': 40,
    '43_instant_noodles': 40,
    '44_instant_noodles': 40,
    '45_instant_noodles': 40,
    '46_instant_noodles': 40,
    '47_instant_noodles': 40,
    '48_instant_noodles': 40,
    '49_instant_noodles': 40,
    '4_puffed_food': 30,
    '50_instant_noodles': 40,
    '51_instant_noodles': 40,
    '52_instant_noodles': 40,
    '53_instant_noodles': 40,
    '54_dessert': 50,
    '55_dessert': 50,
    '56_dessert': 50,
    '57_dessert': 50,
    '58_dessert': 50,
    '59_dessert': 50,
    '5_puffed_food': 30,
    '60_dessert': 50,
    '61_dessert': 50,
    '62_dessert': 50,
    '63_dessert': 50,
    '64_dessert': 50,
    '65_dessert': 50,
    '66_dessert': 50,
    '67_dessert': 50,
    '68_dessert': 50,
    '69_dessert': 50,
    '6_puffed_food': 30,
    '70_dessert': 50,
    '71_drink': 45,
    '72_drink': 45,
    '73_drink': 45,
    '74_drink': 45,
    '75_drink': 45,
    '76_drink': 45,
    '77_drink': 45,
    '78_drink': 45,
    '79_alcohol': 200,
    '7_puffed_food': 30,
    '80_alcohol': 200,
    '81_drink': 45,
    '82_drink': 45,
    '83_drink': 45,
    '84_drink': 45,
    '85_drink': 45,
    '86_drink': 45,
    '87_drink': 45,
    '88_alcohol': 200,
    '89_alcohol': 200,
    '8_puffed_food': 30,
    '90_alcohol': 200,
    '91_alcohol': 200,
    '92_alcohol': 200,
    '93_alcohol': 200,
    '94_alcohol': 200,
    '95_alcohol': 200,
    '96_alcohol': 200,
    '97_milk': 35,
    '98_milk': 35,
    '99_milk': 35,
    '9_puffed_food': 30
}
model1 = YOLO('yolov8n.pt')
BASKET_FRAMES = 25
VOTE_FRAMES = 15

ok, first = cap.read()
rx, ry, rw, rh = cv2.selectROI("Draw BASKET zone then press ENTER", first, showCrosshair=False)
cv2.destroyWindow("Draw BASKET zone then press ENTER")
BASKET = (rx, ry, rx + rw, ry + rh)

history = deque(maxlen=BASKET_FRAMES)
class_votes = defaultdict(lambda: deque(maxlen=VOTE_FRAMES))
basket = Counter()   # current stable basket
events = []  # pickup / put-back log
finalized = False

def in_zone(cx, cy, z):
    return z[0] <= cx <= z[2] and z[1] <= cy <= z[3]


def stable_basket(history):
    # median count per class over the window
    out = Counter()
    for c in set().union(*history):
        vals = sorted(f.get(c, 0) for f in history)
        q = vals[len(vals) // 2]
        if q > 0:
            out[c] = q
    return out
counted = set()
price_bill = 0
purchased_products = []

RADIUS = 300           # px: product must be this close to a person to be linked
OWNER_VOTES = 15        # recent frames used to decide the owner
MIN_VOTE_SHARE = 0.6    # below this the owner is "unsure" (None)
owner_votes = defaultdict(lambda: deque(maxlen=OWNER_VOTES))
person_baskets = defaultdict(Counter)   # {person_id: Counter({class: qty})}

def box_dist(px, py, box):
    x1, y1, x2, y2 = box[:4]
    nx, ny = max(x1, min(px, x2)), max(y1, min(py, y2))
    return ((px - nx) ** 2 + (py - ny) ** 2) ** 0.5

def nearest_person(cx, cy, people, radius=RADIUS):
    best = None
    for p in people:
        d = box_dist(cx, cy, p)
        if d > radius:
            continue
        pcx, pcy = (p[0] + p[2]) / 2, (p[1] + p[3]) / 2
        key = (d, ((cx - pcx) ** 2 + (cy - pcy) ** 2) ** 0.5)
        if best is None or key < best[0]:
            best = (key, p[4])
    return best[1] if best else None

def owner_of(tid, cx, cy, people):
    pid = nearest_person(cx, cy, people)
    if pid is not None:
        owner_votes[tid].append(pid)
    votes = owner_votes[tid]
    if not votes:
        return None
    pid, n = Counter(votes).most_common(1)[0]
    return pid if n / len(votes) >= MIN_VOTE_SHARE else None
while True:
    success,img = cap.read()
    if not success:
        break
    result = model.track(img,stream=True,persist=True,verbose=False)
    result1 = model1.track(img,stream=True,persist=True,classes=[0],verbose=False)

    product = []
    people = []

    for r in result:
        boxes = r.boxes
        for box in boxes:
            x1,y1,x2,y2 = box.xyxy[0]
            x1,y1,x2,y2 = int(x1),int(y1),int(x2),int(y2)
            #print(x1,y1,x2,y2)
            cv2.rectangle(img,(x1,y1),(x2,y2),(255,0,255),2)
            conf = math.ceil((box.conf[0]*100))/100
            #if conf<0.6:
                #continue
            cls = int(box.cls[0])
            className = model.names[cls]
            if box.id is None:
                continue
            track_id = int(box.id[0])


            label = f"{className} {conf}{track_id} "
            cv2.putText(
                img,
                label,
                (x1+5, y1+25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (255, 0, 255),
                2
            )
            product.append((x1,y1,x2,y2,className,track_id))


            '''if track_id not in counted:
                counted.add(track_id)
                price = prices[className]
                price_bill +=price
                purchased_products.append({
                    "track_id": track_id,
                    "product": className,
                    "price": price
                })'''

    for r in result1:
        boxes = r.boxes
        for box in boxes:
            x1,y1,x2,y2 = box.xyxy[0]
            x1,y1,x2,y2 = int(x1),int(y1),int(x2),int(y2)
            #print(x1,y1,x2,y2)
            cv2.rectangle(img,(x1,y1),(x2,y2),(0,0,255),2)
            conf = math.ceil((box.conf[0]*100))/100
            if conf<0.6:
                continue
            if box.id is None:
                continue
            track_id = int(box.id[0])
            label = f"person : {track_id}"
            cv2.putText(img, label, (x1+5, y1+25), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 255), 2)
            people.append((x1, y1, x2, y2, track_id))


    people_by_id = {p[4]: p for p in people}
    frame_counts = Counter()
    zone_owners = defaultdict(Counter)       # class -> who owns the items inside the zone
    for (x1, y1, x2, y2, name, tid) in product:
        class_votes[tid].append(name)
        name = Counter(class_votes[tid]).most_common(1)[0][0]
        cx, cy = (x1 + x2) // 2, (y1 + y2) // 2
        owner = owner_of(tid, cx, cy, people)
        inside = in_zone(cx, cy, BASKET)
        if inside:
            frame_counts[name] += 1
            zone_owners[name][owner] += 1
        if owner in people_by_id:            # draw the link: product -> person
            p = people_by_id[owner]
            cv2.line(img, (cx, cy), ((p[0] + p[2]) // 2, (p[1] + p[3]) // 2), (0, 255, 255), 1)
        color = (0, 255, 0) if inside else (255, 0, 255)
        tag = f"P{owner}" if owner is not None else "?"
        cv2.putText(img, f"{name} -> {tag} {'IN' if inside else 'out'}", (x1, y2 + 20),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)

    history.append(frame_counts)
    new_basket = stable_basket(history)

    zc = ((BASKET[0] + BASKET[2]) // 2, (BASKET[1] + BASKET[3]) // 2)
    for c in set(new_basket) | set(basket):
        d = new_basket.get(c, 0) - basket.get(c, 0)
        if d > 0:
            owner = zone_owners[c].most_common(1)[0][0] if zone_owners[c] else None
            person_baskets[owner][c] += d
            events.append(("PICKUP", owner, c, d, time.strftime("%H:%M:%S")))
            print(f"PICKUP    person {owner}  +{d} {c}")
        elif d < 0:
            holders = [o for o, cnt in person_baskets.items() if cnt.get(c, 0) > 0]
            if holders:
                owner = min(holders, key=lambda o: box_dist(*zc, people_by_id[o]) if o in people_by_id else 1e9)
                person_baskets[owner][c] += d                        # d is negative
                person_baskets[owner] = +person_baskets[owner]       # drop zero entries
                events.append(("PUT_BACK", owner, c, -d, time.strftime("%H:%M:%S")))
                print(f"PUT_BACK  person {owner}  -{-d} {c}")
    basket = new_basket

    price_bill = sum(prices.get(c, 0) * q for c, q in basket.items())
    cv2.rectangle(img, BASKET[:2], BASKET[2:], (0, 255, 255), 2)
    cv2.putText(img, f"Bill: Rs.{price_bill}", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 255), 2)
    y = 60
    for o, cnt in person_baskets.items():
        bill = sum(prices.get(c, 0) * q for c, q in cnt.items())
        if bill:
            cv2.putText(img, f"Person {o}: Rs.{bill}", (10, y),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
            y += 28
    cv2.imshow("IMAGE", img)

    key = cv2.waitKey(1) & 0xFF
    if key == ord('c'):
        finalized = True
        break
    if key == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
machine_output = {
    "items": dict(basket),
    "total": sum(prices.get(c, 0) * q for c, q in basket.items()),
    "people": {o: {"items": dict(cnt),
                   "total": sum(prices.get(c, 0) * q for c, q in cnt.items())}
               for o, cnt in person_baskets.items() if sum(cnt.values()) > 0},
    "finalized": finalized,
    "events": events,
}
print(machine_output)
print("Final basket:")
for c, q in basket.items():
    print(f"  {q}x {c} | Rs.{prices.get(c, 0) * q}")
print("Total: Rs.", price_bill)

print("Events:")
for e in events:
    print(" ", e)
```

```python
def check_pos(camera_input, checkout_data):
    cam = camera_input['items']
    chk = checkout_data['items']

    x = {}   # camera saw more than checkout (not billed)
    y = {}   # checkout has more than camera (ODO · **Hackathon:** TODO · **Demo video:** TODOnot detected)

    for prd in sorted(set(cam) | set(chk)):
        diff = cam.get(prd, 0) - chk.get(prd, 0)
        if diff > 0:
            x[prd] = diff
            print(f"product: {prd} x{diff} added to cart but not present at checkout")
        elif diff < 0:
            y[prd] = -diff
            print(f"product: {prd} x{-diff} at checkout but not detected by camera")

    cam_total = camera_input['total']
    chk_total = checkout_data['total']

    if chk_total > cam_total:
        print(f"camera predicted lower bill than real by {chk_total - cam_total} and didn't detect {y}")
    elif chk_total < cam_total:
        print(f"camera predicted higher bill than checkout by {cam_total - chk_total} and {x} were detected by machine but not present at checkout")
    elif x or y:
        print("TOTALS MATCH but items differ")
    else:
        print("BILL MATCHED")

    return not x and not y
```
